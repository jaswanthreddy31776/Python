⭐ Important Interview Questions
These are the questions I recommend you prepare from this Day 3 file.
Basic Questions
1. What is a Boolean in Python?
A Boolean represents one of two values: True or False.

2. What are the logical operators in Python?
The main logical operators are and, or, and not.

3. What does the and operator do?
It returns True when all the conditions are True.

4. What does the or operator do?
It returns True when at least one condition is True.

5. What does the not operator do?
It reverses the Boolean value. True becomes False and False becomes True.

Assignment Questions
6. What is an assignment operator?
The = operator assigns a value to a variable.

7. What happens here?
a = 10
a = 20

First a gets 10, then it is reassigned to 20. The final value of a is 20.

Increment / Decrement Questions
8. What is an increment operation?
Increment means increasing a value, usually by 1. In Python we use a += 1.

9. What is a decrement operation?
Decrement means decreasing a value, usually by 1. In Python we use a -= 1.

10. Does Python support a++?
No. Python does not use ++ for increment. We use a += 1.

11. What is the difference between these?
a = a + 1


and
a += 1


Both perform the same operation: they increase a by 1. a += 1 is the shorter form.

Type casting is the process of converting a value from one data type to another.

left and right shift Operator

```python
# in left shift "x<<n ==> x*(2**n)"
print(5<<1)
print(5<<2)
print(5<<3)
print(5<<4)
print(5<<5)
```

```python
# in right shift "x>>n ==> x/2**n"
print(100>>2)
print(100>>3)
```

```python
print(100>>4<<3)
#explanation first u have to do right shift and u will get a op value so u have to perfor the left shift with that value.
print(100>>4)
print(6<<3)
```

Special Operators

```python
a=1
b=1
print(id(a))  #id is used to get the memory addres of a variable.
print(id(b))
print(a is b)
print(a==b)
```

```python
a=[1,2,3]
b=[1,2,3]
print(id(a))
print(id(b))
print(a is b)
print(a==b)
```

```python
a=[1,2,3]
b=[1,2,3]
print(id(a))
print(id(b))
print(a is not b)
print(a==b)
```
